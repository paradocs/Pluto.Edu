````
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio

# Function to generate a sine wave for a given frequency and duration
def generate_sine_wave(freq, duration, sample_rate=44100):
    t = np.linspace(0, duration, int(sample_rate * duration), endpoint=False)
    wave = 0.5 * np.sin(2 * np.pi * freq * t)  # Amplitude of 0.5
    return wave

# Generate a sine wave for the note A4 (440 Hz) for 1 second
frequency = 440  # Frequency of A4
duration = 1     # Duration in seconds
sine_wave = generate_sine_wave(frequency, duration)

# Play the generated sine wave
Audio(sine_wave, rate=44100)
````

````
import numpy as np
import matplotlib.pyplot as plt

# Function to create points for a seven-pointed star
def seven_pointed_star():
    # Number of points
    num_points = 7

    # Angles for the star points
    angles = np.linspace(0, 2 * np.pi, num_points, endpoint=False)

    # Radii for the outer and inner points
    outer_radius = 1
    inner_radius = 0.5

    # Create the x and y coordinates
    x = []
    y = []

    for i in range(num_points):
        # Outer point
        x.append(outer_radius * np.cos(angles[i]))
        y.append(outer_radius * np.sin(angles[i]))
        # Inner point
        x.append(inner_radius * np.cos(angles[i] + np.pi / num_points))
        y.append(inner_radius * np.sin(angles[i] + np.pi / num_points))

    return x, y

# Get the star coordinates
x, y = seven_pointed_star()

# Plot the star
plt.figure(figsize=(6, 6))
plt.fill(x, y, color='gold', edgecolor='black')  # Fill with gold color and black edge
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.title('Seven-Pointed Star')
plt.gca().set_aspect('equal')  # Equal aspect ratio
plt.axis('off')  # Turn off the axes
plt.show()
````
